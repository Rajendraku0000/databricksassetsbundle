```python
from pyspark.sql import Row
import random

# Sample data pools
departments = ["Engineering", "Sales", "Marketing", "HR", "Finance"]
first_names = ["Alice", "Bob", "Carla", "David", "Eva", "Frank", "Grace", "Henry", "Ivy", "Jack"]
last_names = ["Johnson", "Smith", "Diaz", "Lee", "Green", "Wright", "Kim", "Ford", "Chen", "Ryan"]

# Generate random employee rows
data = []
for emp_id in range(1, 21):  # 20 employees
    name = f"{random.choice(first_names)} {random.choice(last_names)}"
    dept = random.choice(departments)
    salary = round(random.uniform(50000, 100000), 2)
    hire_year = random.randint(2015, 2024)
    hire_month = random.randint(1, 12)
    hire_day = random.randint(1, 28)
    hire_date = f"{hire_year}-{hire_month:02d}-{hire_day:02d}"
    data.append(Row(emp_id=emp_id, emp_name=name, department=dept, salary=salary, hire_date=hire_date))

# Create DataFrame
df = spark.createDataFrame(data)
df = df.withColumn("hire_date", df["hire_date"].cast("date"))

# Write as managed table inside the schema
df.write.mode("overwrite").saveAsTable("bronze_test.test_schema.emp")


```
